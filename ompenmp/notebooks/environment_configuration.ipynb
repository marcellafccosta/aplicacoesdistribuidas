## 1 - Install Miniconda
* Miniconda is a compact version of Anaconda, that is one of the best package managers to create virtual environments and manage code packages and libraries

#### 1.1 In the terminal and inside your user home, run line-by-line:

* mkdir -p ~/miniconda3

* wget https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh -O ~/miniconda3/miniconda.sh

* bash ~/miniconda3/miniconda.sh -b -u -p ~/miniconda3

* rm -rf ~/miniconda3/miniconda.sh

    -- **Attention**: depends on you SO and Shell configurations:

        -- ~/miniconda3/bin/conda init bash ---> Default for common Linux distributions like Ubuntu or Debian

        -- ~/miniconda3/bin/conda init zsh  ---> Default shell for macOS since Catalin (10,15)

#### 1.2 After the steps above the installation will ask you for restart the terminal

After you restart the terminal, **you will see a text (base)** just before your user name in the terminal.

**For example: (base) daniel@LaPlace:~$**. This means that you are in the base miniconda virtual environment

    -- It is not recommended to install packages in the base environment. 
    -- You should create an especific/dedicated miniconda environment for each project you will work on.

## 2 - Create an Anaconda Virtual Environment for the Classes

In the terminal, execute the following commands (obs.: puc_openmp is only a name for the environment, feel free to use another):

* conda create -n puc_openmp    ---> This will create an empyt environment

* conda activate puc_openmp     ---> This will activate your terminal

To deactivate a virtual environment, execute the following command in the terminal:

* conda deactivate

More about Anaconda: https://docs.conda.io/projects/conda/en/latest/user-guide/tasks/manage-environments.html#sharing-an-environment

### 3.1 - Install the Jupyter Notebook Kernel

By installing the notebook kernel, you will be able to program and execute you code directly from this classes notebooks :-)

You can use Jupyter Notebook with C++ by installing a C++ kernel, such as xeus-cling, which is based on the Cling C++ interpreter. Here's how you can set it up:

In the terminal, with the conda environment activated, execute the following commands

* If you don't already have Jupyter Notebook installed, you can install it using pip:

    * pip install notebook

* Install Xeus-Cling: Xeus-Cling is a Jupyter kernel for C++.

    1.  Add the Conda Forge channel to able conda to find tha packages:
        * conda config --add channels conda-forge
        * conda config --set channel_priority strict

    2. Install Xeus-Cling: This will install the kernel along with the Cling C++ interpreter.
        * conda install xeus-cling -c conda-forge

    3. Verify the Installation: After installation, verify that the C++ kernel is available.
        * jupyter kernelspec list
        * **You should see an entry for xcpp or xeus-cling.**
