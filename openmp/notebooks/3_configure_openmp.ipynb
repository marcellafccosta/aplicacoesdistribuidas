### 4.2.1 -  Identifique os caminhos corretos do GCC

* O kernel Xeus-Cling precisa saber onde estão localizados os arquivos de biblioteca (lib) e cabeçalhos (include) do GCC. 

* **Para encontrar esses caminhos corretos, execute a seguinte linha de código no seu terminal**

    -- gcc -print-search-dirs

* No resultado, procure por:
    * **Libraries** : Diretório contendo libgomp.so (geralmente algo como lib).
    * **Includes**: Diretório contendo os cabeçalhos do GCC (geralmente algo como include).

* Se você seguiu todo o tutorial corretamente até aqui, provavelmente os caminhos para as Libraries e Includes serão como estes:

    * /home/NOME_DO_SEU_USUARIO/miniconda3/envs/puc_openmp/lib

    * /home/NOME_DO_SEU_USUARIO/miniconda3/envs/puc_openmp/include

    * /home/NOME_DO_SEU_USUARIO/miniconda3/envs/puc_openmp/lib/gcc/x86_64-conda-linux-gnu/10.4.0/include/omp.h
    

* Copie os caminhos para as libraries e para os includes na célula abaixo, substituindo os caminhos de exemplo.

* Execute a próxima célula dentro do notebook para permitir que o notebook execute os programas OpemMP corretamente.

## 4.3 Escreva  e teste um programa OpenMP em C++  dentro do Notebook

* Após configurar o kernel, teste o suporte ao OpenMP dentro do notebook executando o código da célula abaixo (4.3.1).

*  **O código deve ser executado sem erros**. 
    * **Porém, você verá que apenas uma thread será executada.**

    * Isso ocorre porque o kernel Xeus-Cling, ao executar o código no Jupyter Notebook, não está configurado para utilizar múltiplos threads por padrão. O comportamento do OpenMP depende de variáveis de ambiente e da configuração do sistema. No terminal, essas variáveis geralmente estão configuradas corretamente, mas no Jupyter Notebook, elas podem precisar ser ajustadas.

    * Desta forma, podemos escrever nossos códigos em C++ e testá-los normalmente mas, para facilitar a execução de várias threads, iremos escrever nosso código em disco e depois executá-lo a partir do terminal.

```c++
#include <iostream>
#include <omp.h>
#include <stdio.h>
int main() {
   #pragma omp parallel
   printf("Hello from thread %d\n", omp_get_thread_num());
   return 0;
}
main();
```
