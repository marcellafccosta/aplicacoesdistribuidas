**Obs: dentro do notebook é executada apenas uma thread. Por isso, execute fora do notebook para ver o paralelismo.** 


O seguinte código executa duas operações sobre os arrays a e b:

```c++
#include "notebooks_reserved_code/openmp_config.h"
#include <omp.h>
#include <stdio.h>
#include <stdlib.h>


int main()
{
	int n = 50;
	int i, nthreads, tid;
	float a[n], b[n], c[n], d[n];

	/* Some initializations */
	for(i=0; i<n; i++){
	  a[i] = i * 1.5;
	  b[i] = i + 22.35;
	  c[i] = d[i] = 0.0;
	 }

	/*1 - O QUE ESTÁ ACONCENDO AQUI?*/
	#pragma omp parallel shared(a,b,c,d,nthreads) private(i,tid)
	{
		tid = omp_get_thread_num();
		/*2 - O QUE ESTÁ ACONCENDO AQUI?*/
		if(tid == 0)
		{
			nthreads = omp_get_num_threads();
			printf("Number of threads = %d\n", nthreads);
		}
		printf("Thread %d starting...\n",tid);
		
		/*3 - O QUE ESTÁ ACONCENDO AQUI?*/
		#pragma omp sections nowait
		{
			/*4 - O QUE ESTÁ ACONCENDO AQUI?*/
			#pragma omp section
			{
			  printf("Thread %d doing section 1\n",tid);
			  for(i=0; i<n; i++){
				  c[i] = a[i] + b[i];
				  printf("Thread %d: c[%d]= %f\n",tid,i,c[i]);
				}
			}
			/*5 - O QUE ESTÁ ACONCENDO AQUI?*/
			#pragma omp section
			{
			  printf("Thread %d doing section 2\n",tid);
			  for(i=0; i<n; i++){
				  d[i] = a[i] * b[i];
				  printf("Thread %d: d[%d]= %f\n",tid,i,d[i]);
				}
			}

		} 

		printf("Thread %d done.\n",tid); 

	}  

}
main()
```

2. Observe que, no código, há comentários numerados, como este: /*1 - O QUE ESTÁ ACONCENDO AQUI?*/. Para cada comentário, explique abaixo o que está acontecendo no respectivo código ao qual o comentário se refere.

    a. Comentário 1:
Está sendo criado um ambiente de paralelismo com a diretiva #pragma omp parallel. Todas as instruções do bloco serão executadas por múltiplas threads simultaneamente. Tem variáveis shared (a, b, c, d, nthreads), que são acessada por todos, e variáveis private (i, tid), que são uma cópia individual por thread. 


    b. Comentário 2:
A thread executando este trecho obtém seu identificador com omp_get_thread_num(), e se for a thread de ID 0 (geralmente a principal), ela armazena na variável nthreads o número total de threads criadas com omp_get_num_threads(), e imprime esse valor na tela.


    c. Comentário 3:
Está utilizando a diretiva #pragma omp sections nowait, que permite que diferentes threads executem blocos de código independentes ao mesmo tempo. Cada seção é atribuída a uma thread disponível, otimizando a execução paralela. O nowait evita que as threads esperem umas pelas outras ao final das seções.

    d. Comentário 4:
Nessa seção, uma das threads executa a soma dos vetores `a` e `b`, armazenando o resultado em `c`.


    e. Comentário 5: 
Nessa seção, uma das threads executa a multiplicação dos vetores `a` e `b`, armazenando o resultado em `d`.
