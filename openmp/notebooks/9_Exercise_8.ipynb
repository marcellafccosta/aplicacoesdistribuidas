O tempo de execução não depende apenas do número de threads, mas também do tamanho do vetor e da sobrecarga de gerenciamento das threads. Para vetores pequenos, aumentar o número de threads pode não trazer ganhos e até aumentar o tempo devido ao overhead. Para vetores maiores, o paralelismo pode ajudar, mas muitas threads podem causar contenção e overhead, limitando a melhora no desempenho. Portanto, é importante balancear o número de threads e o tamanho da tarefa para obter o melhor desempenho possível.
