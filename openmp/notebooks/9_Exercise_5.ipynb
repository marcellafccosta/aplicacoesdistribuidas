**O código abaixo tenta mostrar o uso da construção paralela for.**

**No entanto, ele gerará erros em tempo de compilação.**

**Em seguida, execute conforme as instruções acima.**

````c++
#include "notebooks_reserved_code/openmp_config.h"
#include <omp.h>
#include <stdio.h>
#include <stdlib.h>

int main () 
{
	const int N=50;
	const int CHUNKSIZE=5;
	
	int i, chunk, tid;
	float a[N], b[N], c[N];

	/* Some initializations */
	for (i=0; i < N; i++){
		a[i] = b[i] = i * 1.0;
	}
  
	chunk = CHUNKSIZE;

	#pragma omp parallel for shared(a,b,c,chunk) private(i,tid) schedule(static,chunk)
	{
		tid = omp_get_thread_num();
		for (i=0; i < N; i++){
			c[i] = a[i] + b[i];
			printf("tid= %d i= %d c[i]= %f\n", tid, i, c[i]);
		}
	}  /* end of parallel for construct */

}
main()
````

O tempo de execução depende tanto do tamanho do problema quanto do número de threads. Para problemas pequenos, usar muitas threads pode aumentar o overhead e deixar a execução mais lenta. Já para problemas maiores, usar mais threads tende a melhorar o desempenho. Portanto, é importante equilibrar o número de threads com a carga de trabalho para obter o melhor resultado.


Cole também o código corrigido.

```c
#include "openmp_config.h"
#include <omp.h>
#include <stdio.h>
#include <stdlib.h>

int main()
{
    const int N = 50;
    const int CHUNKSIZE = 5;

    int i, chunk, tid;
    float a[N], b[N], c[N];

    double start_time, end_time;

    /* Some initializations */
    for (i = 0; i < N; i++)
    {
        a[i] = b[i] = i * 1.0;
    }

    chunk = CHUNKSIZE;
    
    omp_set_num_threads(8);
    start_time = omp_get_wtime();

    #pragma omp parallel for shared(a, b, c, chunk) private(i, tid) schedule(static, chunk)
    for (i = 0; i < N; i++)
    {
        tid = omp_get_thread_num();
        c[i] = a[i] + b[i];
        printf("tid= %d i= %d c[i]= %f\n", tid, i, c[i]);
    }
    /* end of parallel for construct */

    end_time = omp_get_wtime();
    printf("Time: %f seconds\n", end_time - start_time);

    return 0;
}
